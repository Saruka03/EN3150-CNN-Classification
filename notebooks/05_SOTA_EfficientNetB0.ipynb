```python
import sys
import os

# Add parent root directory to system path
current_dir = os.getcwd()
parent_dir = os.path.abspath(os.path.join(current_dir, '..'))
if parent_dir not in sys.path:
    sys.path.append(parent_dir)

from src.efficientnet_b0 import create_efficientnet_b0_model

# Initialize EfficientNet-B0 SOTA model
efficientnet_model = create_efficientnet_b0_model()
efficientnet_model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist

# 1. Load Fashion-MNIST data
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# 2. Train the model
history = efficientnet_model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

# 3. Evaluate on the test set
test_loss, test_accuracy = efficientnet_model.evaluate(X_test, y_test)
print(f"EfficientNet-B0 Test Accuracy: {test_accuracy:.4f}")

# 4. Save the trained model to the root 'models/' folder
os.makedirs(os.path.join('..', 'models'), exist_ok=True)
model_save_path = os.path.join('..', 'models', 'Model_EfficientNetB0_Thayarathan.keras')
efficientnet_model.save(model_save_path)
print(f"Model saved successfully to {model_save_path}")
```
